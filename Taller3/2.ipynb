```python
import numpy as np
import scipy.signal as signal
import matplotlib.pyplot as plt
import wave

# Definición de las funciones de filtrado

def filtro_pasabajo(x, f_corte, fs):
    taps = signal.firwin(151, f_corte, fs=fs)
    filtered_x = signal.lfilter(taps, 1.0, x)
    return filtered_x

def filtro_pasaalto(x, f_corte, fs):
    taps = signal.firwin(151, f_corte, pass_zero=False, fs=fs)
    filtered_x = signal.lfilter(taps, 1.0, x)
    return filtered_x

def filtro_pasabanda(x, f_corte1, f_corte2, fs):
    taps = signal.firwin(151, [f_corte1, f_corte2], pass_zero=False, fs=fs)    
    filtered_x = signal.lfilter(taps, 1.0, x)
    return filtered_x

# Parámetros de la señal
fm = 48000   # Frecuencia de muestreo (Hz)
T = np.arange(0, 1, 1/fm)  # Tiempo en segundos (1 segundo de duración)
# Composición de una señal con 4 frecuencias
Y = np.sin(2 * np.pi * 1000 * T) + np.sin(2 * np.pi * 5000 * T) + np.sin(2 * np.pi * 10000 * T) + np.sin(2 * np.pi * 15000 * T)

# Aplicar los filtros

# Filtro pasa bajo (ejemplo, corte en 5000 Hz)
f_corte_pasabajo = 5000
Y_pasabajo = filtro_pasabajo(Y, f_corte_pasabajo, fm)

# Filtro pasa alto (ejemplo, corte en 5000 Hz)
f_corte_pasaalto = 5000
Y_pasaalto = filtro_pasaalto(Y, f_corte_pasaalto, fm)

# Filtro pasa banda (ejemplo, corte entre 5000 Hz y 10000 Hz)
f_corte_pasabanda1 = 5000
f_corte_pasabanda2 = 10000
Y_pasabanda = filtro_pasabanda(Y, f_corte_pasabanda1, f_corte_pasabanda2, fm)

# Graficar resultados

plt.figure(figsize=(10, 6))

# Señal original
plt.subplot(3, 1, 1)
plt.plot(T[:1000], Y[:1000])  # Graficar solo los primeros 1000 puntos para una visualización clara
plt.title("Señal Original (combinación de frecuencias)")
plt.xlabel("Tiempo [s]")
plt.ylabel("Amplitud")

# Señal filtrada con pasa bajo
plt.subplot(3, 1, 2)
plt.plot(T[:1000], Y_pasabajo[:1000])
plt.title(f"Señal Filtrada Pasa Bajo (Corte {f_corte_pasabajo} Hz)")
plt.xlabel("Tiempo [s]")
plt.ylabel("Amplitud")

# Señal filtrada con pasa alto
plt.subplot(3, 1, 3)
plt.plot(T[:1000], Y_pasaalto[:1000])
plt.title(f"Señal Filtrada Pasa Alto (Corte {f_corte_pasaalto} Hz)")
plt.xlabel("Tiempo [s]")
plt.ylabel("Amplitud")

plt.tight_layout()
plt.show()
```

# CONVOLUCION

Debido a esta naturaleza “deslizante”, la longitud de la salida es en realidad más larga que la de la entrada. Si una señal tiene M muestras y la otra señal es N muestras, La convolución de los dos puede producir N+M-1 muestras. Sin embargo, funciones como numpy.convolve() tiene una manera de especificar si desea la salida completa (max(M, N) muestras) o simplemente las muestras donde las señales se superpusieron completamente (max(M, N) - min(M, N) + 1 si eres curioso). No hay necesidad de quedar atrapado en este detalle. Solo sepa que la longitud de la salida de una convolución no es solo la longitud de las entradas.
