```python
import numpy as np
import matplotlib.pyplot as plt

years=[2006 +x for x in range(16)]
weights=[80, 83, 84, 85, 86, 82, 81, 79, 83, 80, 82, 82, 83, 81, 80,79]

plt.plot(years,weights)
#puntos conectados
plt.show()
#añado el show para q se cree una nueva figura en scatter a continuacion
#sino se superpondrian las figuras


plt.scatter(years,weights)
plt.show()
#puntos sin conexion

plt.plot(years,weights, c="red", lw=1, linestyle="--")
```

Histogramas

```python
import numpy as np
import matplotlib.pyplot as plt

ages=np.random.normal(20, 1.5, 1000)
#normal distribution, (mean, standar desviation, size)

plt.hist(ages,
         bins=[ages.min(), 18,21, ages.max()],
         color="black")
plt.show()

plt.hist(ages,
         bins=200,
         cumulative=True,
         color="black")

```

Pie Charms

```python
import matplotlib.pyplot as plt

langs=["C++","C#","Python","Java","GO"]
votes=[20,30,100,40,69]
plt.pie(votes, labels=langs)
plt.show()
#pie tiene sentido si la funcion es inyectiva

explodes=[0.2,0,0,0,0]
plt.pie(votes, labels=langs,explode=explodes, #explode es para separar un dato 
        autopct="%.2f%%") #pra hallar los % aqui 0.2 son dos decimales 
#ademas con pctdistance= 1.2 distancia de los pocertanjes del centro
#startangule =90,lo mueve, lo rota
```

Multiples Figuras

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import style

x1,y1 = np.random.random(100),np.random.random(100)
x2,y2= np.arange(100),np.arange(100)

plt.figure(1)
plt.scatter(x1,y1)

plt.figure(2)
plt.plot(x2,y2)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import style

plt.style.use("default")
x= np.arange(100)

fig, axs=plt.subplots(2,2)

axs[0,0].plot(x, np.sin(x))
axs[0,0].set_title("Sine Wave")

axs[1,0].plot(x, np.cos(x))
axs[1,0].set_title("Cosine Wave")

axs[0,1].plot(x, np.log(x))
axs[0,1].set_title("Log Function")

axs[1,1].plot(x, np.random.random(100))
axs[1,1].set_title("Random")
axs[1,1].set_xlabel("Ñ")

plt.suptitle("Functions")
#plt.title no funciona
```

3D Ploting

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import style

ax=plt.axes(projection="3d")
x=np.random.random(100)
y=np.random.random(100)
z=np.random.random(100)

ax.plot(x,y,z)
ax.set_title("3d plot")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import style

ax=plt.axes(projection="3d")
x=np.arange(0,50,0.1)
y=np.sin(x)
z=np.cos(x)

ax.plot(x,y,z)
ax.set_title("3d plot")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import style

ax=plt.axes(projection="3d")
x=np.arange(0,10,0.1)
y=np.arange(0,10,0.1)
X,Y =np.meshgrid(x,y)
Z=np.sin(X)*np.cos(Y)

ax.plot_surface(X,Y,Z)
ax.set_title("3d plot")
plt.show()
```

Animar la funcion

```python
import numpy as np
import matplotlib.pyplot as plt
import random

heads_tails =[0,0]
for i in range(100000):
    heads_tails[random.randint(0,1)] += 1
    plt.bar(["Heads","Tails"], heads_tails, color=["blue","red"])
    plt.pause(0.001)
plt.show()
```
