```python
import numpy as np



def promedio_movil(vector, N):
    prom=0
    pm=np.zeros((vector.size - N + 1)) #se crea un array lleno de 0s pero de dimensión necesitada para guardar los valores del promedio movil

    for i in range((vector.size - N + 1)): #Este ciclo recorre el indice de i hasta el ultimo i posible, que generaliza la inecuacion dada
        for j in range(N):                 #Este ciclo recorre las N para hallar el promedio
            prom=(prom*j +vector[i+j])/(j+1) #Aqui se usa la ecuacion de promedio general dada en el ejercicio anterior,
                                             #pero en el ar[i+j]  se usa i para que vaya tirando otro promedio a medida que i avance,
                                             # y conseguir nuevos promedios para cada vez que el indice i avance
        pm[i]=prom                          #se reescribe el array pm llenandolo con los promedios moviles
    return pm







#FUNCION AUXILIAR PARA GRAFICAR
import numpy as np
import matplotlib.pyplot as plt
def graficar(ve,N,titulo):
  # matriz es un array(M,N) donde M es la cantidad de
  # vectores a graficar y N es el largo de cada vector
    f, ax = plt.subplots(1, 1, figsize=(10,10))
    ax.plot(ve)
    ax.set_xlim((0, N))
    ax.set_ylim((0, 1))
    #ax.set_xlabel('mi etiqueta x')
    #ax.set_ylabel('mi etiqueta y')
    ax.set_title(titulo)
    plt.tight_layout()
    
    #plt.show()
    plt.savefig('gráfico_1.pdf')


# código de prueba de la función anterior
vector = np.random.random(1000)
for N in range(1,21):
    pm_movil = promedio_movil(vector, N)
    # pintamos resultado
    graficar(pm_movil, np.size(pm_movil), titulo="Ventana movil, N="+ str(N))
```
