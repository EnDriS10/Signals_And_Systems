```python
import numpy as np

a = input("Introduce un numero")
A = np.zeros((int(a),int(a)))
A[2,2]=-2
A[0,1]=1
A[1,0]=1
b=int(a)

lista_entrada =input(f"Intrduzca una lista de numeros que contenga {b} elementos que acaba de introducir, y separados por ,")
lista= [float(i) for i in lista_entrada.split(",")]
v=np.array([lista])

print(A @ v.reshape((int(a),1))) #reshape para cambiarlo a una matriz vertical

```
