```python
# Inicializar una lista vacía
mi_lista = []

# Preguntar cuántos elementos quiere agregar
n = int(input("¿Cuántos elementos deseas agregar a la lista? "))

# Agregar elementos a la lista
for i in range(n):
    elemento = input(f"Ingrese el elemento {i + 1}: ")
    mi_lista.append(elemento)

# Imprimir la lista completa
print("La lista creada es:", mi_lista)
```

```python
#funcion que multiplique y sume vectores

u =[2.0,3.4]
v = [1,9.8]

if (len(u) == len(v)):
    suma = []
    for i in range (len(u)):
        a = u[i]+ v[i]
        suma.append(a)
    print(suma)
else:
    print("Las listas deben tener las mismas dimensiones")

#Para multiplicar

if (len(u) == len(v)):
    multiplicacion = []
    for j in range (len(u)):
        b = u[j] * v[j]
        multiplicacion.append(b)
    print(multiplicacion)
else:
    print("Las listas deben tener las mismas dimensiones")


```

```python
#elevar una lista a la otra con funciones

a=[2.3,3, 4]
b=[2, 4, 4]


def Elevar_Listas(x,y):
    result = []
    if len(x) == len(y):  
        for i in range(len(x)):
            c = x[i]**y[i]
            result.append(c)
    else:
        print("Las listas deben tener el mismo largo")
    return result

print(Elevar_Listas(a,b))

```
